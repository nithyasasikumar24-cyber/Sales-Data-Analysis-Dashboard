```python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact *= i

    return fact

print("Factorial =", factorial(5))
```

```python
def count_vowels(text):
    count = 0

    for ch in text.lower():
        if ch in "aeiou":
            count += 1

    return count

word = input("Enter a word: ")

print("Number of vowels =", count_vowels(word))
```

```python
marks = []

for i in range(5):
    mark = int(input(f"Enter mark {i+1}: "))
    marks.append(mark)

# Calculate total manually
total = 0
for mark in marks:
    total = total + mark

average = total / len(marks)

# Find highest mark manually
highest = marks[0]
for mark in marks:
    if mark > highest:
        highest = mark

# Find lowest mark manually
lowest = marks[0]
for mark in marks:
    if mark < lowest:
        lowest = mark

print("\n----- Student Marks Report -----")
print("Marks:", marks)
print("Total:", total)
print("Average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
```
