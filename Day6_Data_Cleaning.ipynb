```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Nithya", "Rahul", "Anu", "Arjun", "Meera"],
    "Age": [21, np.nan, 20, 23, np.nan],
    "Department": ["CSE", "ECE", "CSE", None, "CSE"],
    "Marks": [95, 88, np.nan, 79, 85]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Department"] = df["Department"].fillna("Unknown")
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
df.dropna()
```

```python
df.drop_duplicates()
```

```python
df.rename(columns={"Marks": "Score"}, inplace=True)

df
```

```python
print(df.info())
print(df.describe())
```

```python
import pandas as pd
import numpy as np

employee = {
    "Employee": ["Alex", "John", "John", "Sara"],
    "Salary": [50000, np.nan, 45000, 60000],
    "Department": ["HR", "IT", "IT", None]
}

emp = pd.DataFrame(employee)

print(emp)

print(emp.isnull().sum())

emp["Salary"] = emp["Salary"].fillna(emp["Salary"].mean())

emp["Department"] = emp["Department"].fillna("Unknown")

emp = emp.drop_duplicates()

print(emp)
```
