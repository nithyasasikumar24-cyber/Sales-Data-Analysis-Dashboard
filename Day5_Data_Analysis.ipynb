```python
import pandas as pd

data = {
    "Name": ["Nithya", "Rahul", "Anu", "Arjun", "Meera"],
    "Age": [21, 22, 20, 23, 21],
    "Department": ["CSE", "ECE", "CSE", "EEE", "CSE"],
    "Marks": [95, 88, 91, 79, 85]
}

df = pd.DataFrame(data)

df
```

```python
print(df.head())
print(df.tail())
print(df.shape)
print(df.columns)
```

```python
print(df.info())
print(df.describe())
```

```python
print(df[["Name", "Marks"]])
```

```python
high_marks = df[df["Marks"] > 90]
print(high_marks)
```

```python
print(df.sort_values(by="Marks"))
```

```python
df["Result"] = ["Pass", "Pass", "Pass", "Pass", "Pass"]

print(df)
```

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("Total Students:", df["Name"].count())
```

```python
df.to_csv("students_day5.csv", index=False)
```
