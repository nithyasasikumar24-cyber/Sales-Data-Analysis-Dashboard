```python
df = pd.read_csv("processed_sales_data.csv")
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("===== SALES DASHBOARD =====")

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
print("Total Records:", df["Sales"].count())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month_name()
```

```python
monthly_sales = df.groupby("Month", sort=False)["Sales"].sum()

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales.plot(kind="bar", figsize=(8,5))

plt.title("Product-wise Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python
# Sales Dashboard Summary

## Dashboard Components

- Total Sales
- Average Sales
- Maximum Sales
- Minimum Sales
- Product-wise Sales Bar Chart
- Monthly Sales Trend
- Sales Contribution Pie Chart

## Conclusion

The dashboard provides a clear overview of sales performance by presenting key business metrics and visualizations. It helps identify product performance, monthly trends, and sales distribution, making it easier to support business decision-making.
```
