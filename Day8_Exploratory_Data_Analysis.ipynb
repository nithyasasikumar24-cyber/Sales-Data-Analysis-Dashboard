```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Product": [
        "Laptop", "Mobile", "Tablet", "Laptop",
        "Mobile", "Headphones", "Watch", "Tablet"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Accessories",
        "Accessories", "Electronics"
    ],
    "Sales": [
        55000, 42000, 28000, 60000,
        45000, 18000, 24000, 30000
    ],
    "Profit": [
        10000, 8000, 5000, 12000,
        9000, 3500, 4500, 5500
    ],
    "Quantity": [5, 8, 6, 6, 9, 12, 10, 7]
}

df = pd.DataFrame(data)

df
```

```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Highest Sale:", df["Sales"].max())
print("Lowest Sale:", df["Sales"].min())
print("Total Profit:", df["Profit"].sum())
```

```python
df.loc[df["Sales"].idxmax()]
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

print(product_sales)
```

```python
product_sales.plot(kind="bar")

plt.title("Product-wise Total Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
category_sales.plot(kind="bar")

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True)

plt.show()
```

```python
correlation = df[["Sales", "Profit", "Quantity"]].corr()

correlation
```

```python
top_sales = df.nlargest(3, "Sales")

top_sales
```
