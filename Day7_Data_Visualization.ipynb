```python
import pandas as pd

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 13500, 18000, 21000, 19500],
    "Profit": [2500, 3200, 2800, 4200, 5100, 4600]
}

df = pd.DataFrame(data)

df
```

```python
plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.plot(df["Month"], df["Sales"], marker="o", label="Sales")
plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")

plt.title("Monthly Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

```python
plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
categories = ["Electronics", "Clothing", "Food", "Furniture"]
sales = [40, 25, 20, 15]

plt.pie(
    sales,
    labels=categories,
    autopct="%1.1f%%"
)

plt.title("Sales by Category")

plt.show()
```

```python
sales_data = [
    12000, 15000, 13500, 18000, 21000,
    19500, 17000, 16000, 22000, 14500
]

plt.hist(sales_data, bins=5)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
products = ["Laptop", "Mobile", "Tablet", "Headphones", "Watch"]
product_sales = [55000, 42000, 28000, 18000, 24000]

plt.bar(products, product_sales)

plt.title("Product-wise Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```
