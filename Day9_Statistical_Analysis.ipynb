```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Headphones", "Watch",
                "Laptop", "Mobile", "Tablet", "Headphones", "Watch"],
    
    "Sales": [55000, 42000, 28000, 18000, 24000,
              62000, 47000, 31000, 20000, 27000],
    
    "Profit": [10000, 8000, 5000, 3500, 4500,
               12500, 9000, 5800, 3800, 4900],
    
    "Quantity": [5, 8, 6, 12, 10, 6, 9, 7, 14, 11]
}

df = pd.DataFrame(data)

df
```

```python
df.describe()
```

```python
print("Mean Sales:", df["Sales"].mean())
print("Median Sales:", df["Sales"].median())
print("Mode Sales:", df["Sales"].mode()[0])
```

```python
print("Minimum Sales:", df["Sales"].min())
print("Maximum Sales:", df["Sales"].max())

print("Minimum Profit:", df["Profit"].min())
print("Maximum Profit:", df["Profit"].max())
```

```python
print("Sales Variance:", df["Sales"].var())
print("Sales Standard Deviation:", df["Sales"].std())
```

```python
print("25th Percentile:", df["Sales"].quantile(0.25))
print("50th Percentile:", df["Sales"].quantile(0.50))
print("75th Percentile:", df["Sales"].quantile(0.75))
```

```python
correlation = df[["Sales", "Profit", "Quantity"]].corr()

correlation
```

```python
plt.hist(df["Sales"], bins=5)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.grid(True)

plt.show()
```

```python
average_sales = df.groupby("Product")["Sales"].mean()

print(average_sales)
```

```python
average_sales.plot(kind="bar")

plt.title("Average Sales by Product")
plt.xlabel("Product")
plt.ylabel("Average Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
average = df["Sales"].mean()

print("Overall Average Sales:", average)
```

```python
above_average = df[df["Sales"] > average]

above_average
```
